```python
## Sampling rate
sr = 407.25
## Identify the non-zero entries of the peaks array
non_zero_indices = np.nonzero(peaks)
## Form a dictionary where for each node index, we have a list of times where peaks occurred
peaks_dict = {}
for time_idx, node_idx in zip(non_zero_indices[0], non_zero_indices[1]):
    if node_idx not in peaks_dict:
        peaks_dict[int(node_idx)] = []
    peaks_dict[int(node_idx)].append(float(time_idx))
## Sort the lists of times for each node, and add random uniform noise to each time point
np.random.seed(42)  # For reproducibility
for node_idx in peaks_dict:
    ## peaks_dict[node_idx] = [(time + np.random.uniform(0, 1)) / (sr * 60) for time in peaks_dict[node_idx]]
    peaks_dict[node_idx] = [time / (sr * 60) for time in peaks_dict[node_idx]]
    peaks_dict[node_idx].sort()
```

```python
## Save peaks dictionary as a npy file
np.save('../data/sea_slug/Dec0922_peak_times.npy', peaks_dict)
```

Sep1223: This file, taken by my colleague, Viral Mistry, is also a concatenation of several acquisitions--5 of them, in this case, lasting 4 minutes each. Whereas my Jun0425 experiment was meant to subtract learning from the equation, Viral's directly leverages it, as only 10 minutes elapsed between successive acquisitions. Note that, in contrast to the preceding three files, the sampling rate in this file is 1.629 kHz. The stims are at 0.5, 4.5, 8.5, 12.5, and 16.5 minutes (time steps 48870, 439830, 830790, 1221750, and 1612710), while the breaks between the five constituent acquisitions are at 4, 8, 12, and 16 minutes (390960, 781920, 1172880, 1563840). All stimuli were of equal intensity.

```python
## Import peaks from npy file
import numpy as np
peaks = np.load('../data/sea_slug/Sep1223_peaks.npy')
```

```python
def adjust_time(t, sr):
    if t < 390960:
        return t
    elif t < 781920:
        return t + sr * 60 * 10 
    elif t < 1172880:
        return t + 2 * (sr * 60 * 10)
    elif t < 1563840:
        return t + 3 * (sr * 60 * 10)
    else:
        return t + 4 * (sr * 60 * 10)
```

```python
## Sampling rate
sr = 1629.0
## Identify the non-zero entries of the peaks array
non_zero_indices = np.nonzero(peaks)
## Form a dictionary where for each node index, we have a list of times where peaks occurred
peaks_dict = {}
for time_idx, node_idx in zip(non_zero_indices[0], non_zero_indices[1]):
    if node_idx not in peaks_dict:
        peaks_dict[int(node_idx)] = []
    peaks_dict[int(node_idx)].append(float(time_idx))
## Sort the lists of times for each node, and add random uniform noise to each time point
for node_idx in peaks_dict:
    peaks_dict[node_idx] = [adjust_time(time, sr) / (sr * 60) for time in peaks_dict[node_idx]]
    peaks_dict[node_idx].sort()
```

```python
## Save peaks dictionary as a npy file
np.save('../data/sea_slug/Sep1223_peak_times.npy', peaks_dict)
```
